## Variables, types, and conversion

A variable is a name that stores a value. In Python, a variable can store different types of values.
Use clear names for variables. You can use letters, numbers, and underscores, but a variable name cannot start with a number

Common built-in types are `str`, `int`, `float`, `bool`, and `None` (the absence of a value). `type()` helps inspect a value. Convert values explicitly with constructors such as `int()` and `float()`.

````python
student_name = "Ali"
age = 20
height_m = 1.75
is_student = True
middle_name = None

print(type(student_name), type(age), type(height_m), type(is_student), type(middle_name))

text_number = "10"
integer_number = int(text_number)
decimal_number = float(text_number)
print(integer_number + 5, decimal_number / 2)

# Multiple assignment
x, y, z = 1, 2, 3
print(x, y, z)
````

## Strings

An f-string embeds expressions directly inside text. Useful methods include `upper()`, `lower()`, `strip()`, `replace()`, and `split()`.

## Conditions and Boolean logic

## Loops

Use a `for` loop to visit each item in an iterable. `range(start, stop, step)` produces integer sequences; its stop value is excluded. `enumerate()` supplies both an index and an item. A `while` loop repeats while its condition remains true—update the condition to avoid an infinite loop. Use `break` to exit early.

````python
for number in range(1, 6):
    print(number, end=" ")
print()

fruits = ["apple", "banana", "cherry"]
for index, fruit in enumerate(fruits, start=1):
    print(index, fruit)

for row in range(2):
    for column in range(3):
        print(f"({row}, {column})", end=" ")
    print()

count = 0
while count < 3:
    count += 1
    print(f"Attempt {count}")

print(list(range(5, 0, -1)))

for city in ["Cairo", "Alexandria", "quit", "Luxor"]:
    if city == "quit":
        break
    print(city)
````

## Lists

A list is an ordered, mutable collection. Use indexing and slicing to access values, assignment to replace an item, and methods such as `append()`, `extend()`, `insert()`, `pop()`, `remove()`, and `del` to change it. `+=` extends a list in place. `sort()` and `reverse()` change the list; `sorted()` returns a sorted copy.

Copy a list with `[:]` (or `.copy()`) when you need an independent list. Comprehensions offer a compact way to build a new list. Numeric lists work well with `min()`, `max()`, and `sum()`.

````python
names = ["mahmoud", "ali", "ahmed"]
names[1] = "sara"
names.append("mohamed")
names.extend(["rana", "nour"])
names.insert(1, "basma")
removed_name = names.pop()
names.remove("ahmed")
names += ["salma"]
del names[0]
print(names, "removed:", removed_name)

first_three = names[:3]
copied_names = names[:]
copied_names.append("independent")
print(first_three, copied_names)

numbers = [5, 3, 1, 4, 2]
print(sorted(numbers), numbers)
numbers.sort()
print(numbers)
numbers.reverse()
print(numbers)
print(min(numbers), max(numbers), sum(numbers))

squares = [number ** 2 for number in range(1, 6)]
print(squares)
````

## Tuples, sets, and dictionaries

A **tuple** is an ordered, immutable collection. A one-item tuple needs a trailing comma; convert a tuple to a list when you need a changed copy. A **set** is an unordered collection of unique items, useful for removing duplicates and set operations. `discard()` removes a value without failing when it is absent. A **dictionary** maps unique keys to values. Prefer `get()` when a key may not exist.

````python
dimensions = (200, 50)
single_item = (5,)
print(dimensions[0], type(single_item))
updated_dimensions = list(dimensions)
updated_dimensions[0] = 250
print(tuple(updated_dimensions))

unique_scores = {90, 75, 90, 60}
unique_scores.add(100)
unique_scores.discard(60)  # No error if the value is absent.
print(unique_scores)
print({1, 2, 3}.union({3, 4}), {1, 2, 3}.intersection({3, 4}))

student = {"name": "Yasmine", "age": 20, "languages": ["Python", "SQL"]}
student["country"] = "Egypt"
student["age"] = 21
del student["country"]
print(student.get("email", "not provided"))
for key, value in student.items():
    print(f"{key}: {value}")
print(list(student.keys()), list(student.values()))
````

## Functions

Functions package reusable behavior. Define one with `def`, then call it by name. Parameters receive values supplied as arguments. Positional arguments are matched by order; keyword arguments are matched by name. Default parameter values make an argument optional.

`return` sends a result back to the caller. A function without `return` implicitly returns `None`. Docstrings briefly describe a function's purpose.

````python
def describe_pet(animal_type, pet_name="Max"):
    """Return a short description of a pet."""
    return f"My {animal_type} is named {pet_name.title()}."


def full_name(first_name, last_name):
    return f"{first_name} {last_name}".title()


print(describe_pet("dog"))
print(describe_pet(pet_name="Luna", animal_type="cat"))
print(full_name("ola", "hassan"))


def greet_users(usernames):
    for username in usernames:
        print(f"Hello, {username.title()}!")


greet_users(["ali", "samy", "mahmoud"])
````

## Scope and mutable arguments

A variable created inside a function is local to that function. A function can read a global value, but assigning to it creates a local name unless `global` is used. Usually, return a new value instead of changing global state.

Lists are mutable objects. Passing a list gives a function access to the same object, so in-place changes are visible to the caller. Copy the list first when that is not wanted.

````python
TAX_RATE = 0.24

def price_with_tax(price):
    return price * (1 + TAX_RATE)


def add_priority(tasks):
    tasks.append("review")


tasks = ["plan", "write"]
add_priority(tasks)
print(price_with_tax(100), tasks)

unchanged_tasks = tasks[:]
add_priority(unchanged_tasks)
print("original:", tasks)
print("copied:", unchanged_tasks)
````

## Recursion and lambda functions

A recursive function calls itself and needs a **base case** that stops further calls. Recursion is useful for naturally recursive problems, although a loop is often simpler for basic counting tasks.

A `lambda` is a small anonymous function containing one expression. Name a lambda only when that makes code clearer; use `def` for larger behavior.

````python
def factorial(number):
    """Return number! for a non-negative integer."""
    if number < 0:
        raise ValueError("factorial is undefined for negative numbers")
    if number <= 1:
        return 1
    return number * factorial(number - 1)


raise_power = lambda base, exponent: base ** exponent

print(factorial(5))
print(raise_power(2, 3))
````

###  Task: Bank System

Write a Python program for a simple bank system that:

1. Create a variable `balance` with an initial value of **1000**.
2. Create three functions:

   * `check_balance()` → displays the current balance.
   * `deposit()` → asks the user for an amount and adds it to the balance.
   * `withdraw()` → asks the user for an amount and subtracts it from the balance.
3. Use a **`while` loop** to display a menu repeatedly.
4. The menu should contain:

   * `1` → Check Balance
   * `2` → Deposit
   * `3` → Withdraw
   * `4` → Exit
5. Use **`if/elif/else`** to handle the user's choice.
6. In the withdrawal function, use `if/else` to check whether the balance is sufficient.
7. Display a suitable message after each operation.


````python
balance=1000
def check_balance():
    print("balance",balance)
def deposit():
    global balance
    amount=int(input("enter the amount:"))
    balance+=amount
    print("deposit successfull")
def withdraw():
    global balance
    amount = float(input("Enter withdrawal amount: "))
    if amount <= balance:
        balance = balance - amount
        print("Withdrawal successful.")
    else:
        print("Insufficient balance.")
while True:
    print("\n1. Check Balance")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Exit")
    choice = input("Choose an option: ")
    
    if choice == "1":
        check_balance()
    elif choice == "2":
        deposit()
    elif choice == "3":
        withdraw()
    elif choice == "4":
        print("Thank you!")
        break
    else:
        print("Invalid choice.")
````

# Object-Oriented Programming (OOP) 

## 1. Procedural Programming vs OOP

### **1.1 Procedural Programming (Traditional Code)**

Procedural programming is based on functions and step-by-step execution.
Characteristics:

* Data and functions are **separate**.
* Code is executed in a linear, sequential flow.
* Reusability is limited — often requires copy-pasting functions.
* Suitable for small programs.

````python
# Procedural example
name = "Ahmed"
age = 20
grade = "A"
def print_student(name, age, grade):
    print(f"Name: {name}, Age: {age}, Grade: {grade}")

print_student(name, age, grade)
````

### **1.2 Object-Oriented Programming (OOP)**

OOP organizes code around **objects** that combine:

* **Attributes** (Data)
* **Methods** (Functions on that data)

Characteristics:

* Data + behavior are grouped inside a **class**.
* Code is reusable and scalable.
* Excellent for medium/large projects.

**Example (OOP Version):**

````python
class Student:
    def __init__(self,name,age,grade):
        self.name=name
        self.age=age
        self.grade=grade
    def print_info(self):
        print(f"\nName:{self.name},Age:{self.age},Grade:{self.grade}")
    
    
````

````python
student1 = Student("Ahmed", 20, "A")
student2 = Student("Mahmoud", 23, "B+")
student3 = Student("Alaa", 25, "A+")
````

````python
student1.print_info()
student2.print_info()
student3.print_info()
````

### 2.2 OOP Version (Without Inheritance)

````python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def print_info(self):
        print(f"Title: {self.title}, Author: {self.author}")
````

````python
class Library:      # nested class
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def show_books(self):
        for book in self.books:
            book.print_info()
````

#  **Encapsulation — Definition**

**Encapsulation** means **grouping data (variables) and methods (functions) into one unit**, usually a **class** in Object-Oriented Programming (OOP).

It also means **restricting direct access** to some data to protect it from unwanted changes — like putting data inside a “capsule”.

So encapsulation does two things:

1. **Bundling data + methods together in one class**
2. **Controlling access** using:

   * private attributes (e.g., `__balance`)
   * getters and setters


### **Types of Attributes**

1. **Public**

   * Can be accessed anywhere
   * Example: `self.balance`

2. **Protected** (single underscore `_`)

   * Intended for internal use (child classes can access)
   * Example: `self._balance`

3. **Private** (double underscore `__`)

   * Not accessible outside the class directly
   * Example: `self.__pin`

---

---

# ✅ **Why Encapsulation Is Important**

* Protects data from being changed accidentally.
* Defines **how data should be accessed** (via methods).
* Makes the code cleaner and easier to maintain.
* Increases security.

---

# ✅ **Simple Example (Python): Bank Account**

### ❌ Without Encapsulation (unsafe)

````python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance  # public — anyone can change it

account = BankAccount(1000)
account.balance = -5000   # Someone can break the system!
print(account.balance)
````

# 🔹 **Example: Bank Account with Public, Protected, Private**

````python
class BankAccount:
    def __init__(self, name, balance, pin):
        self.name = name          # public
        self._balance = balance   # protected
        self.__pin = pin          # private

    # Public method
    def deposit(self, amount):
        if amount > 0:
            self._balance += amount
        return self._balance

    # Public method
    def withdraw(self, amount, pin):
        if pin == self.__pin:
            if 0 < amount <= self._balance:
                self._balance -= amount
                return self._balance
            else:
                return "Insufficient funds"
        else:
            return "Invalid PIN"

    # Getter for private attribute
    def get_pin(self):
        return self.__pin
````

````python
# Create object
acc = BankAccount("Ahmed", 1000, 1234)

# Access public attribute
print(acc.name)  # Ahmed
acc.name = "Ali"
print(acc.name)  # Ali

# Access protected attribute
print(acc._balance)  # 1000 (can access but should not outside class)
acc._balance += 500
print(acc._balance)  # 1500

# Access private attribute
# print(acc.__pin)  # Error: AttributeError
print(acc.get_pin())  # 1234 (safe access through getter)

# Use methods
print(acc.deposit(500))         # 2000
print(acc.withdraw(1000, 1234)) # 1000
print(acc.withdraw(1000, 1111)) # Invalid PIN
````



### ✅ With Encapsulation (safe)

We use **private variables** (`__balance`) + **getter** + **setter**.

```python

````python

class BankAccount:
    def __init__(self, balance):
        self.__balance = balance   # private variable

    # Getter (read-only)
    def get_balance(self):
        return self.__balance
    
    # Setter (controlled update)
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
    
    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount

# Usage
account = BankAccount(1000)

print(account.get_balance())  # 1000

account.deposit(500)
print(account.get_balance())  # 1500

account.withdraw(200)
print(account.get_balance())  # 1300
````


**What Happened Here?**

| Concept              | Meaning                                         |
| -------------------- | ----------------------------------------------- |
| **Private variable** | `__balance` cannot be accessed directly         |
| **Getter method**    | Safely read the balance                         |
| **Setter methods**   | Deposit/withdraw with validation                |
| **Encapsulation**    | Data + methods in one class + controlled access |

# Procedural / Functional Banking System

````python
accounts = {}  # global dictionary
def create_account(account_number, name, balance):
    accounts[account_number] = {
        "name": name,
        "balance": balance
    }

def deposit(account_number, amount):
    accounts[account_number]["balance"] += amount


def withdraw(account_number, amount):
    if accounts[account_number]["balance"] >= amount:
        accounts[account_number]["balance"] -= amount
    else:
        print("Insufficient funds!")

def show_account(account_number):
    acc = accounts[account_number]
    print(f"Account: {account_number}, Name: {acc['name']}, Balance: {acc['balance']}")
````

# 🟢 **OOP Version — With Encapsulation, Inheritance, Polymorphism**

## **Base Class: Account (Parent)**

````python
class Account:
    def __init__(self, account_number, name, balance):
        self._account_number = account_number   # protected (encapsulation)
        self._name = name
        self._balance = balance

    def deposit(self, amount):
        self._balance += amount

    def withdraw(self, amount):
        if self._balance >= amount:
            self._balance -= amount
        else:
            print("Insufficient funds!")

    def show_info(self):
        print(f"[Account {self._account_number}] {self._name} → Balance: {self._balance}")
````

# 🟣 **Child Class 1: SavingsAccount (Inheritance + Polymorphism)**

Savings accounts earn interest → **override deposit**.

```python


````python
class SavingsAccount(Account):
    def deposit(self, amount):
        # Polymorphism: different behavior
        interest = amount * 0.02  # 2% interest
        total = amount + interest
        print(f"Deposit with interest: +{interest}")
        self._balance += total
````

# 🟠 **Child Class 2: BusinessAccount (Inheritance + Polymorphism)**

Business accounts charge a fee on withdrawals.

```python


````python
class BusinessAccount(Account):
    def withdraw(self, amount):
        fee = 10
        total = amount + fee
        
        if self._balance >= total:
            print(f"Business withdrawal fee charged: {fee}")
            self._balance -= total
        else:
            print("Insufficient funds for business withdrawal!")
````
