## Variables, types, and casting

## Strings

An f-string drops expressions into a template. Methods you will use often: `upper()`, `lower()`, `strip()`, `replace()`, and `split()`.

````python
first = "lina"
last = "qassem"
banner = f"Hello, {first.title()} {last.title()}!"
print(banner)
word = "kitchen"
print(word[1:5], word[:4], word[4:], word[::-1])
messy = "  Bake bread slowly.  "
tidied = messy.strip().replace("slowly", "with care")
print(tidied)
print("warm crusty loaves".split())
print("e" in word, "q" not in word)
print("Stations:\n\tMixer\n\tOven")
````

A `#` starts a short comment. Python will not freeze a name for you, but writing it in UPPER_CASE signals “please treat this as a constant.”

````python
print(9 + 4, 9 * 4, 3 ** 4)
print(53 / 4, 53 // 4, 53 % 4)
print((8 - 3) * 4)
MAX_ORDERS = 2_400  # Uppercase name: intended as a constant.
print(MAX_ORDERS)
points = 21
points += 9
print(points)
````

## Loops

`for` walks through an iterable one item at a time. `range(start, stop, step)` yields integers and never includes `stop`. `enumerate()` pairs each item with an index. `while` keeps going as long as its condition stays true, so you must update that condition or the loop never ends. `break` jumps out immediately.

````python
for n in range(3, 13, 3):
    print(n, end=" ")
print()
for index, stall in enumerate(["Spice", "Citrus", "Honey"], start=1):
    print(index, stall)
for row in range(2):
    for col in range(3):
        print(f"({row}, {col})", end=" ")
    print()
lap = 0
while lap < 3:
    lap += 1
    print(f"Lap {lap}")
print(list(range(12, 0, -3)))
for city in ["Luxor", "Qena", "halt", "Minya"]:
    if city == "halt":
        break
    print(city)
````

## Lists

A list keeps items in order and can be changed. Reach items with indexes and slices, overwrite with assignment, and reshape with `append()`, `extend()`, `insert()`, `pop()`, `remove()`, and `del`. `+=` grows the list in place. `sort()` and `reverse()` edit the list; `sorted()` builds a new ordered list.

When you need a second independent list, copy with `[:]` or `.copy()`. A comprehension is a compact way to build a list. For numbers, `min()`, `max()`, and `sum()` are available.

````python
bakers = ["amal", "youssef", "dalia"]
bakers[1] = "nabil"
bakers.append("reem")
bakers.extend(["karim", "salwa"])
bakers.insert(1, "farah")
dropped = bakers.pop()
bakers.remove("dalia")
bakers += ["tamer"]
del bakers[0]
print(bakers, "dropped:", dropped)
head = bakers[:3]
clone = bakers[:]
clone.append("extra")
print(head, clone)
scores = [19, 5, 14, 7, 3]
print(sorted(scores), scores)
scores.sort(); print(scores)
scores.reverse(); print(scores)
print(min(scores), max(scores), sum(scores))
print([n ** 3 for n in range(1, 6)])
````

## Tuples, sets, and dictionaries

A **tuple** is ordered and cannot be mutated. A tuple of one value needs a trailing comma. Convert to a list if you must change items. A **set** ignores order and stores each value once, which is handy for uniqueness and set algebra. `discard()` removes a value if it is present and stays quiet if it is not. A **dictionary** associates unique keys with values. Prefer `get()` when a key might be missing.

````python
frame = (720, 480)
solo = (19,)
print(frame[0], type(solo))
editable = list(frame)
editable[0] = 1080
print(tuple(editable))
tickets = {"open", "queued", "open", "done"}
tickets.add("hold")
tickets.discard("done")
print(tickets)
print({1, 3, 5}.union({5, 7}), {1, 3, 5}.intersection({5, 7}))
staff = {"name": "Tariq", "shift": 1, "skills": ["Mixing", "Proofing"]}
staff["city"] = "Tanta"; staff["shift"] = 2; del staff["city"]
print(staff.get("email", "missing"))
for key, value in staff.items(): print(f"{key}: {value}")
print(list(staff.keys()), list(staff.values()))
````

## Functions

A function is a named block you can reuse. Write it with `def` and run it by calling that name. Parameters catch the values you pass in. Order matches **positional** arguments; names match **keyword** arguments. Defaults let a caller skip some arguments.

`return` sends a result back. If you never `return`, the function yields `None`. A docstring is a short note describing what the function is for.

````python
def introduce_animal(kind, animal_name="Nimbus"):
    """Build a one-line description of an animal."""
    return f"The {kind} is named {animal_name.title()}."

def format_name(given, family):
    return f"{given} {family}".title()

print(introduce_animal("otter"))
print(introduce_animal(animal_name="Pebble", kind="fox"))
print(format_name("hani", "mostafa"))

def greet_group(people):
    for person in people: print(f"Hello, {person.title()}!")
greet_group(["tariq", "lina", "omar"])
````

## Scope and mutable arguments

A name you create inside a function stays local. The function may read a global name, but assigning to that name makes a new local unless you declare `global`. Returning a result is usually clearer than rewriting globals.

Because lists can change, in-place edits inside a function show up for the caller. Copy first if you want the original left alone.

````python
TAX_RATE = 0.12
def price_with_tax(amount): return amount * (1 + TAX_RATE)
def push_step(steps): steps.append("pack")
steps = ["mix", "bake"]
push_step(steps)
print(price_with_tax(50), steps)
steps_copy = steps[:]
push_step(steps_copy)
print("original:", steps)
print("copy:", steps_copy)
````

## Recursion and lambda functions

A recursive function calls itself. It must include a **base case** so the chain of calls eventually stops. Recursion fits problems that split into similar smaller problems; a loop is often easier for plain counting.

A `lambda` is an unnamed function with a single expression. Reach for `def` once the logic grows.

````python
def factorial(number):
    """Compute factorial for a non-negative integer."""
    if number < 0: raise ValueError("factorial is undefined for negatives")
    if number <= 1: return 1
    return number * factorial(number - 1)
power_of = lambda base, exponent: base ** exponent
print(factorial(5))
print(power_of(3, 4))
````

### Task: tiny till

Set `balance` to **420**. Write functions that display it, add cash, and pay out. Drive them from a repeating `while` menu with choices `1` to `4`. Use `if/elif/else` to pick an action, and refuse a payment that would go past the remaining money.

````python
balance = 420
def check_balance(): print("Current balance:", balance)
def add_money():
    global balance
    balance += float(input("Amount to add: "))
    print("Cash added.")
def spend_money():
    global balance
    amount = float(input("Amount to pay: "))
    if amount <= balance:
        balance -= amount; print("Payment recorded.")
    else: print("Not enough cash.")
def run_till():
    while True:
        print("\n1. Check balance\n2. Add money\n3. Spend money\n4. Exit")
        choice = input("Choose an option: ")
        if choice == "1": check_balance()
        elif choice == "2": add_money()
        elif choice == "3": spend_money()
        elif choice == "4": print("See you later!"); break
        else: print("Pick one of the numbered options.")
# run_till()
````

# Object-oriented programming (OOP)

## 1. Procedures versus objects

### 1.1 Procedural programming

A procedural program is a sequence of functions and steps. Data lives in one place and functions live in another. That split is fine for small scripts, but it often leads to copied logic as the project expands.

````python
runner_name = "Malak"
runner_age = 22
runner_band = "C"
def print_runner(name, age, band):
    print(f"Name: {name}, Age: {age}, Band: {band}")
print_runner(runner_name, runner_age, runner_band)
````

### 1.2 Object-oriented programming

OOP builds programs from **objects**. Each object holds **attributes** (its data) and **methods** (functions that use that data). Keeping those together encourages reuse and often makes larger programs easier to grow.

````python
class Runner:
    def __init__(self, name, age, band):
        self.name = name
        self.age = age
        self.band = band

    def print_info(self):
        print(f"Name: {self.name}, Age: {self.age}, Band: {self.band}")
````

````python
runner_one = Runner("Malak", 22, "C")
runner_two = Runner("Tarek", 28, "A")
runner_three = Runner("Nourhan", 25, "B")
````

## 4. A short look at polymorphism

````python
class Publication:
    def print_info(self):
        print("This is a general publication")

class PaperBook(Publication):
    def print_info(self):
        print("Paper book with pages")

class DigitalBook(Publication):
    def print_info(self):
        print("Digital book with a file size")

publications = [PaperBook(), DigitalBook(), Publication()]
for publication in publications:
    publication.print_info()
````

### Method overriding

**Overriding** means a child class defines a method with the same name as one on the parent, then gives it new behavior. Polymorphism relies on this idea.

````python
class Pass:
    def show_type(self):
        print("Day pass")

class SeasonPass(Pass):
    def show_type(self):
        print("Season pass with guest tickets")

basic = Pass()
season = SeasonPass()
basic.show_type()
season.show_type()
````

Both classes expose `show_type`, yet the printed text depends on which object you call it on.

### Method overloading

In languages such as Java and C++, you can declare several methods that share a name but take different parameters. Classic overloading is not a Python feature, but default arguments can mimic the convenience.

````python
class Calculator:
    def add(self, first, second, third=0):
        return first + second + third

calculator = Calculator()
print(calculator.add(4, 12))
print(calculator.add(4, 12, 6))
````

One `add` method handles two numbers or three because the last parameter has a default.

## Encapsulation

**Encapsulation** keeps related data and methods inside a class and decides how that data may be changed. If an attribute is fully public, outside code can assign nonsense to it. Public names are meant for general use, a leading `_` is a convention for “protected,” and a leading `__` makes a name private via name mangling. Methods can then check values before they change anything.

````python
class OpenTill:
    def __init__(self, balance):
        self.balance = balance

till = OpenTill(650)
till.balance = -80  # Nothing stops this invalid assignment.
print(till.balance)
````

### Example: public, protected, and private fields

````python
class SecureTill:
    def __init__(self, owner, balance, pin):
        self.owner = owner             # public
        self._balance = balance        # protected by convention
        self.__pin = pin               # private

    def deposit(self, amount):
        if amount > 0:
            self._balance += amount
        return self._balance

    def withdraw(self, amount, pin):
        if pin != self.__pin:
            return "Incorrect pin"
        if 0 < amount <= self._balance:
            self._balance -= amount
            return self._balance
        return "Insufficient balance"

    def get_pin(self):
        return self.__pin
````

````python
till = SecureTill("Yasmin", 980, 4172)
print(till.owner)
till.owner = "Hala"
print(till.owner)
print(till._balance)
till._balance += 120
print(till._balance)
# print(till.__pin)  # AttributeError
print(till.get_pin())
print(till.deposit(220))
print(till.withdraw(400, 4172))
print(till.withdraw(80, 9999))
````

### Controlled access with encapsulation

Hiding the balance behind methods that read and update it lets the class validate every change.

````python
class GuardedTill:
    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount

till = GuardedTill(650)
print(till.get_balance())
till.deposit(190)
print(till.get_balance())
till.withdraw(75)
print(till.get_balance())
````

| Idea | Meaning |
| --- | --- |
| Private variable | `__balance` is not read from outside |
| Getter method | Returns the balance in a controlled way |
| Update methods | Check deposits and withdrawals first |
| Encapsulation | Data and the operations on it stay together |

## Part 1 — a procedural ledger

This version stores every account in one global dictionary and uses functions to change it.

````python
accounts = {}
def create_account(number, owner, balance):
    accounts[number] = {"owner": owner, "balance": balance}
def deposit(number, amount):
    accounts[number]["balance"] += amount
def withdraw(number, amount):
    if accounts[number]["balance"] >= amount:
        accounts[number]["balance"] -= amount
    else:
        print("Insufficient balance!")
def show_account(number):
    account = accounts[number]
    print(f"Account: {number}, Owner: {account['owner']}, Balance: {account['balance']}")
````

### Usage

````python
create_account(512, "Hala", 3100)
deposit(512, 480)
withdraw(512, 250)
show_account(512)
````

## Part 2 — limits of the function-based design

Anyone can reach the global dictionary and rewrite a balance. Invalid values are easy to sneak in. New account kinds often mean copying functions, and there is no natural place for different deposit or withdrawal rules.

## Part 3 — the same idea with objects

Here, encapsulation keeps each account’s data on the object, inheritance reuses the base account, and polymorphism lets child classes change selected operations.

````python
class Account:
    def __init__(self, number, owner, balance):
        self._number = number
        self._owner = owner
        self._balance = balance

    def deposit(self, amount):
        self._balance += amount

    def withdraw(self, amount):
        if self._balance >= amount:
            self._balance -= amount
        else:
            print("Insufficient balance!")

    def show_info(self):
        print(f"[Account {self._number}] {self._owner}: {self._balance}")
````

### Child class 1: savings account

A savings account overrides `deposit()` so a small bonus is added.

````python
class SavingsAccount(Account):
    def deposit(self, amount):
        bonus = amount * 0.04
        print(f"Deposit bonus: {bonus}")
        self._balance += amount + bonus
````

### Child class 2: business account

A business account overrides `withdraw()` so a fee is charged.

````python
class BusinessAccount(Account):
    def withdraw(self, amount):
        fee = 20
        total = amount + fee
        if self._balance >= total:
            print(f"Business withdrawal fee: {fee}")
            self._balance -= total
        else:
            print("Insufficient balance for this withdrawal!")
````

### Driving the OOP ledger

````python
savings = SavingsAccount(821, "Malak", 3400)
business = BusinessAccount(822, "Karim", 7500)
savings.deposit(600)
savings.withdraw(280)
savings.show_info()
business.withdraw(900)
business.deposit(350)
business.show_info()
````

## OOP recap

| Concept | In this example |
| --- | --- |
| Encapsulation | Each object stores its own account fields |
| Inheritance | `SavingsAccount` and `BusinessAccount` build on `Account` |
| Polymorphism | Children replace deposit or withdrawal behavior |
| Organization | Actions live next to the data they use |

The procedural sketch depends on a shared dictionary. The object-oriented version keeps state with each account, adds specialized types without copying everything, and is simpler to extend.

# Object-oriented programming (OOP)

## Why use OOP?

A machine-learning project rarely stays as a handful of variables and functions. Loading data, cleaning it, building features, fitting, scoring, predicting, and plotting all have to cooperate. OOP is one way to bundle related values and actions into an **object**.

## Classes and objects

A **class** is the plan. An **object** is one living copy of that plan. `Dataset` might be the class; `train_set` and `test_set` would be two objects with their own contents.

```python
class Dataset:
    pass

train_set = Dataset()
test_set = Dataset()
```

## Attributes, `__init__`, and `self`

Attributes are the pieces of data an object remembers, such as a name or a batch size. `__init__` runs as soon as you create the object. `self` is that object, so `self.name = name` stores the argument on this particular instance.

```python
class Dataset:
    def __init__(self, name, batch_size):
        self.name = name
        self.batch_size = batch_size

images = Dataset("Street Photos", 32)
reviews = Dataset("Product Comments", 16)
```

Both objects come from the same class; only their stored values differ.

## Methods

A function defined on a class is a **method**. Methods are the actions the object can take.

```python
class Dataset:
    def __init__(self, name):
        self.name = name

    def shuffle(self):
        print(f"{self.name} is shuffling")

    def batch(self):
        print(f"{self.name} is building a batch")
```

That pairing of **data and behavior** is the heart of OOP.

## OOP in machine learning libraries

You already meet this pattern in libraries. In scikit-learn, `LinearRegression()` constructs an object; `model.fit(...)` and `model.predict(...)` call methods on it. Creating a model and asking it to train or predict is everyday OOP.

## Encapsulation

Encapsulation puts related values and operations in one class — for example a balance together with deposit and withdrawal methods. The class can then hide or check data instead of exposing every field.

## Inheritance

A child class can reuse methods and attributes from a parent.

```python
class BaseEstimator:
    def fit(self):
        print("Fitting")

    def predict(self):
        print("Predicting")

class ImageEstimator(BaseEstimator):
    pass
```

`ImageEstimator` inherits `fit()` and `predict()` without rewriting them. In deep learning, `class CustomModel(nn.Module):` works the same way: your class stands on a PyTorch parent.

## Polymorphism

Polymorphism is one method name with different behavior depending on the object. Two estimator classes can both define `predict()`, while one labels pictures and the other labels sentences.

```python
class ImageEstimator(BaseEstimator):
    def predict(self):
        print("Predicting an image label")

class TextEstimator(BaseEstimator):
    def predict(self):
        print("Predicting the next token")
```

## Putting it together

Classes are reusable designs, objects are created from them, attributes hold data, and methods do work. Children inherit shared behavior and override a method when they need a different action. As an ML codebase grows, classes help you group datasets, preprocessors, models, trainers, and evaluators.

### Quick checks

- In `estimator = ImageEstimator()`, `ImageEstimator` is the class and `estimator` is the object.
- In `estimator.predict()`, `predict()` is a method.
- In `class ImageEstimator(BaseEstimator):`, the child inherits from the parent.
- Separate `predict()` implementations are polymorphism.

You do not need to memorize the vocabulary for its own sake. When code creates an object, calls a method, or subclasses another type, map that back to how real ML libraries are structured.
