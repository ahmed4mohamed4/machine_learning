# Python Basics

## Getting started

Python is widely used in AI and data work because its readable syntax lets you focus on problem solving. A Jupyter notebook contains **Markdown cells** for explanation and **code cells** for executable Python. Run the selected cell with `Shift+Enter`.

`print()` displays values. Strings may use single or double quotes; use the quote style that avoids escaping an apostrophe in the text.

```python
print("Hello, Python!")
print("It\'s easy to choose the other quote style when needed.")
print("Python", "is", "readable")
print("Same line:", end=" ")
print("continued")
print("A multiline string:\nfirst line\nsecond line")
```

## Variables, types, and conversion

A variable is created by assignment and refers to a value. Python is dynamically typed, so a name can later refer to a value of another type, though clear programs usually avoid unnecessary type changes. Use descriptive `snake_case` names; names may contain letters, digits, and underscores, but cannot start with a digit.

Common built-in types are `str`, `int`, `float`, `bool`, and `None` (the absence of a value). `type()` helps inspect a value. Convert values explicitly with constructors such as `int()` and `float()`.

```python
student_name = "Ali"
age = 20
height_m = 1.75
is_student = True
middle_name = None

print(type(student_name), type(age), type(height_m), type(is_student), type(middle_name))

text_number = "10"
integer_number = int(text_number)
decimal_number = float(text_number)
print(integer_number + 5, decimal_number / 2)

# Multiple assignment
x, y, z = 1, 2, 3
print(x, y, z)
```

## Strings

A string is a sequence of characters inside quotes. Strings are **immutable**: methods such as `replace()` produce a new string rather than changing the original. Indexing starts at `0`; negative indices count from the end. Slices use `start:stop:step`, with the stop position excluded.

An f-string embeds expressions directly inside text. Useful methods include `upper()`, `lower()`, `strip()`, `replace()`, and `split()`.

```python
first_name = "mahmoud"
last_name = "hassan"
full_name = f"Hello, {first_name.title()} {last_name.title()}!"
print(full_name)

language = "python"
print(language[1:4], language[:3], language[3:], language[::-1])

messy_text = "  I love Python  "
clean_text = messy_text.strip().replace("Python", "programming")
print(clean_text)
print("I love Python".split())
print("y" in language, "z" not in language)
print("Languages:\n\tPython\n\tJavaScript")
```

## Numbers, operators, comments, and constants

Integers have no decimal part; floats do. Arithmetic uses `+`, `-`, `*`, `/`, `//` (floor division), `%` (remainder), and `**` (power). Division with `/` always returns a float. Parentheses make precedence explicit.

Use `#` for concise comments. Python does not enforce constants, but UPPER_CASE names communicate values intended to stay unchanged.

```python
print(2 + 3, 2 * 3, 2 ** 3)
print(65 / 3, 65 // 3, 65 % 3)
print((1 + 2) * 4)

MAX_CONNECTIONS = 5_000  # Convention: treat this as a constant.
print(MAX_CONNECTIONS)

total = 10
total += 5
print(total)
```

## Input

`input()` always returns a string. Convert it before arithmetic or numeric comparisons. The examples below are defined as functions so the notebook can run without waiting for input; call them when you want to practise interactive input.

`split()` separates one line into pieces, and `map()` applies a conversion function to each piece.

```python
def greet_from_input():
    name = input("Enter your name: ")
    print(f"Hello, {name.title()}!")


def add_three_numbers_from_input():
    values = map(float, input("Enter three numbers separated by spaces: ").split())
    print(f"Total: {sum(values)}")


# Try interactively:
# greet_from_input()
# add_three_numbers_from_input()
```

## Conditions and Boolean logic

An `if` statement runs code when its condition is true. Use `==` to compare values and `=` to assign a value. Combine alternatives with `elif` and `else`. Membership operators `in` and `not in` test whether a value occurs in a collection.

Boolean values are `True` and `False`. Use `and`, `or`, and `not` to combine or invert conditions.

```python
age = 19
if age < 13:
    category = "child"
elif age < 20:
    category = "teenager"
else:
    category = "adult"
print(category)

allowed_users = ["ali", "sara", "mona"]
username = "sara"
print(username in allowed_users)
print(username != "admin")
print(is_student and age >= 18, not username == "admin")
```

## Loops

Use a `for` loop to visit each item in an iterable. `range(start, stop, step)` produces integer sequences; its stop value is excluded. `enumerate()` supplies both an index and an item. A `while` loop repeats while its condition remains true—update the condition to avoid an infinite loop. Use `break` to exit early.

```python
for number in range(1, 6):
    print(number, end=" ")
print()

fruits = ["apple", "banana", "cherry"]
for index, fruit in enumerate(fruits, start=1):
    print(index, fruit)

for row in range(2):
    for column in range(3):
        print(f"({row}, {column})", end=" ")
    print()

count = 0
while count < 3:
    count += 1
    print(f"Attempt {count}")

print(list(range(5, 0, -1)))

for city in ["Cairo", "Alexandria", "quit", "Luxor"]:
    if city == "quit":
        break
    print(city)
```

## Lists

A list is an ordered, mutable collection. Use indexing and slicing to access values, assignment to replace an item, and methods such as `append()`, `extend()`, `insert()`, `pop()`, `remove()`, and `del` to change it. `+=` extends a list in place. `sort()` and `reverse()` change the list; `sorted()` returns a sorted copy.

Copy a list with `[:]` (or `.copy()`) when you need an independent list. Comprehensions offer a compact way to build a new list. Numeric lists work well with `min()`, `max()`, and `sum()`.

```python
names = ["mahmoud", "ali", "ahmed"]
names[1] = "sara"
names.append("mohamed")
names.extend(["rana", "nour"])
names.insert(1, "basma")
removed_name = names.pop()
names.remove("ahmed")
names += ["salma"]
del names[0]
print(names, "removed:", removed_name)

first_three = names[:3]
copied_names = names[:]
copied_names.append("independent")
print(first_three, copied_names)

numbers = [5, 3, 1, 4, 2]
print(sorted(numbers), numbers)
numbers.sort()
print(numbers)
numbers.reverse()
print(numbers)
print(min(numbers), max(numbers), sum(numbers))

squares = [number ** 2 for number in range(1, 6)]
print(squares)
```

## Tuples, sets, and dictionaries

A **tuple** is an ordered, immutable collection. A one-item tuple needs a trailing comma; convert a tuple to a list when you need a changed copy. A **set** is an unordered collection of unique items, useful for removing duplicates and set operations. `discard()` removes a value without failing when it is absent. A **dictionary** maps unique keys to values. Prefer `get()` when a key may not exist.

```python
dimensions = (200, 50)
single_item = (5,)
print(dimensions[0], type(single_item))
updated_dimensions = list(dimensions)
updated_dimensions[0] = 250
print(tuple(updated_dimensions))

unique_scores = {90, 75, 90, 60}
unique_scores.add(100)
unique_scores.discard(60)  # No error if the value is absent.
print(unique_scores)
print({1, 2, 3}.union({3, 4}), {1, 2, 3}.intersection({3, 4}))

student = {"name": "Yasmine", "age": 20, "languages": ["Python", "SQL"]}
student["country"] = "Egypt"
student["age"] = 21
del student["country"]
print(student.get("email", "not provided"))
for key, value in student.items():
    print(f"{key}: {value}")
print(list(student.keys()), list(student.values()))
```

## Functions

Functions package reusable behavior. Define one with `def`, then call it by name. Parameters receive values supplied as arguments. Positional arguments are matched by order; keyword arguments are matched by name. Default parameter values make an argument optional.

`return` sends a result back to the caller. A function without `return` implicitly returns `None`. Docstrings briefly describe a function's purpose.

```python
def describe_pet(animal_type, pet_name="Max"):
    """Return a short description of a pet."""
    return f"My {animal_type} is named {pet_name.title()}."


def full_name(first_name, last_name):
    return f"{first_name} {last_name}".title()


print(describe_pet("dog"))
print(describe_pet(pet_name="Luna", animal_type="cat"))
print(full_name("ola", "hassan"))


def greet_users(usernames):
    for username in usernames:
        print(f"Hello, {username.title()}!")


greet_users(["ali", "samy", "mahmoud"])
```

## Scope and mutable arguments

A variable created inside a function is local to that function. A function can read a global value, but assigning to it creates a local name unless `global` is used. Usually, return a new value instead of changing global state.

Lists are mutable objects. Passing a list gives a function access to the same object, so in-place changes are visible to the caller. Copy the list first when that is not wanted.

```python
tax_rate = 0.14


def price_with_tax(price):
    return price * (1 + tax_rate)


def add_priority(tasks):
    tasks.append("review")


tasks = ["plan", "write"]
add_priority(tasks)
print(price_with_tax(100), tasks)

unchanged_tasks = tasks[:]
add_priority(unchanged_tasks)
print("original:", tasks)
print("copied:", unchanged_tasks)
```

## Modules and files

A module is a Python file whose definitions can be imported elsewhere. In a project, save reusable functions in a file such as `greetings.py`, then import it with `import greetings` and call `greetings.greet_users(...)`.

Use `with open(...)` to work with files: it closes the file automatically, even if an error occurs. The example uses a temporary file so it does not leave files in the notebook directory.

```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as temporary_directory:
    file_path = Path(temporary_directory) / "data.txt"
    file_path.write_text("A short note about Python.", encoding="utf-8")
    contents = file_path.read_text(encoding="utf-8")
    print(contents)
```

## Recursion and lambda functions

A recursive function calls itself and needs a **base case** that stops further calls. Recursion is useful for naturally recursive problems, although a loop is often simpler for basic counting tasks.

A `lambda` is a small anonymous function containing one expression. Name a lambda only when that makes code clearer; use `def` for larger behavior.

```python
def factorial(number):
    """Return number! for a non-negative integer."""
    if number < 0:
        raise ValueError("factorial is undefined for negative numbers")
    if number <= 1:
        return 1
    return number * factorial(number - 1)


raise_power = lambda base, exponent: base ** exponent

print(factorial(5))
print(raise_power(2, 3))
```
